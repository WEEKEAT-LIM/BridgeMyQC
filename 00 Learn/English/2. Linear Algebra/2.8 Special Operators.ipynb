# Special Operators
## Complex Matrix/Vector

### Dagger
We use $\dagger$ (Dagger) to represent complex conjugate transpose, therefore when $z \in \mathbb{C}$:
$$
\begin{align}
A &= \begin{bmatrix} z_1 & z_2 \\ z_3 & z_4 \end{bmatrix} \\
A^\dagger &= \begin{bmatrix} z_1^* & z_2^* \\ z_3^* & z_4^* \end{bmatrix}^T \\
A^\dagger &= \begin{bmatrix} z_1^* & z_3^* \\ z_2^* & z_4^* \end{bmatrix}
\end{align}
$$
Example:
$$
\begin{align}
A &= \begin{bmatrix} 1+2i & 2+3i \\ 1-9i & 10+100i \end{bmatrix} \\
A^\dagger &= \begin{bmatrix} 1-2i & 2-3i \\ 1+9i & 10-100i \end{bmatrix}^T \\
A^\dagger &= \begin{bmatrix} 1-2i & 1+9i \\ 2-3i & 10-100i \end{bmatrix}
\end{align}
$$
> $A^{\dagger\dagger} =A$

### Length
You already know the length of **a complex number**(vector in complex plane) is $\sqrt{z^*z}$, my question is **what is the length of vector $\vec{v}$ if the elements of vector is complex number?** (if you just simply find square root of dot product of itself you 100% get wrong result, you will get complex number again, but the length must be real scalar). Actually it share the same logic with real elements vector.

$\vec{v}_{real} = \begin{bmatrix}a \\b \end{bmatrix} \quad a,b\in \mathbb{R} \quad ||\vec{v}_{real} || = \sqrt{a^2 + b^2}$, $a$ also consider as is horizontal length of the vector and $b$ is vertical length of the vector. In this case we use pythagoras theorem to find the length, because the length is quite obvious.

Actually complex vector still implement 'pythagoras theorem', the different thing is, you need to find the length of each side first before apply pythagoras.

In general:
$$
||\vec{v}_{any \, space}|| = \sqrt{\sum (length \, of \, each \, component)^2}
$$

Hence:
$$
||\vec{v}|| = \sqrt{\vec{v}^\dagger\vec{v}} \quad , \vec{v} \in \mathbb{C}
$$

### Complex Dot Product (Hermitian Inner Product)
$$
\vec{v}^\dagger\vec{u}
$$

> IF You still want to use classical dot product, you must find the length/magnitude of final complex number in complex plane. Then take the square root of the length.

```python
import numpy as np
```

```python
A = np.array([
    [1+3j],
    [2+6j]
])

dotA = A.T@A # Result type: Complex Number
complex_dotA = A.conj().T@A  # Result type: Real Number

print(f"False Length: {np.sqrt(dotA)}\nTrue Length:{np.sqrt(complex_dotA)}={np.sqrt(abs(dotA))}")
```

## Hermitian Operator
$$
H^\dagger = H
$$
Example:
$$
\begin{align}
H_1^\dagger &=\begin{bmatrix}2 & 1 - i \\ 1+i & 3 \end{bmatrix}^\dagger \\
&=  \begin{bmatrix}2 & 1 + i \\ 1-i & 3 \end{bmatrix}^T \\
&= \begin{bmatrix}2 & 1 - i \\ 1+i & 3 \end{bmatrix}
\end{align}
$$
What is Eigenvalues and Eigenvectors of $H_1$?
$$
\begin{align}
Hx &= \lambda x \\
(H-\lambda I)x &= 0 \\
\det (H-\lambda I) &= 0 \\
\det (\begin{bmatrix} 2-\lambda & 1 - i \\ 1+i & 3-\lambda\end{bmatrix}) &= 0 \\
(2-\lambda)(3-\lambda) - (1-i)(1+i) &= 0 \\
\lambda^2 -5\lambda +6 - (1-i^2) &= 0\\
\lambda^2 -5\lambda +4 &=0 \\
(\lambda-4)(\lambda-1) &= 0 \\
\lambda &\in \set{1, 4} \\
\begin{bmatrix} 1 & 1 - i \\ 1+i & 2\end{bmatrix}x_1 = 0 &\quad \begin{bmatrix} -2 & 1 - i \\ 1+i & -1\end{bmatrix}x_2 =0 \\
x_1 = \begin{bmatrix}-1+i \\1 \end{bmatrix} &\quad x_2 = \begin{bmatrix} 1 \\ 1+i\end{bmatrix}
\end{align}
$$
Did you see something interesting:
- In $2 \times 2$ matrix, if the off-diagonal elements are complex conjugate of each other the matrix must be hermitian
- $\lambda \in \mathbb{R}$
- Eigenvectors are orthogonal to each others

Check orthogonal:
$$
\begin{align}
x_1^\dagger x_2 &= \begin{bmatrix}-1-i &1 \end{bmatrix}\begin{bmatrix} 1 \\ 1+i\end{bmatrix} \\
&= -1 -i +1 + i \\
&= 0
\end{align}
$$

```python
H1 = np.array([
    [2, complex(1, -1)],
    [complex(1, 1), 3]
])
print(f"H1 Dagger:\n{H1.conj().T}")

lambdas, eig_vec = np.linalg.eig(H1)
print(f"x1 dagger times x2: {eig_vec[:, 0].conj().T @ eig_vec[:,1]}") # Approximate 0, therefore they are orthogonal
print(f"Eigenvalues: {lambdas}")
```

## Unitary Operator
$$
U^\dagger U = UU^\dagger = I = \begin{bmatrix} 1&0 \\ 0&1\end{bmatrix} \\
U^\dagger = U^{-1}
$$
- $| \det U| = 1$
- **Multiplicative Closure**: Product of any two unitary matrix is also a Unitary Matrix
- **Norm Preservation**: Keeps the Euclidean length of any complex vector unchanged ($||Ux|| = ||x||$) 
- **Inner Product**: Preserve inner product of vectors ($(Ux) \cdot (Uy) = x \cdot y$)
- **Eigenvalues**: All eigenvalues have an absolute value (modulus) of exactly 1 ($|λ| = 1$).

Example:
$$
U = \frac{1}{\sqrt{3}}\begin{bmatrix}1&1-i\\1+i&-1 \end{bmatrix} \quad U^\dagger = \frac{1}{\sqrt{3}}\begin{bmatrix}1&1-i\\1+i&-1 \end{bmatrix}
$$
> In this case $U$ also Hermitian operator, BUT NOT ALL unitary matrices are hermitian

```python
U = np.sqrt(1/3) * np.array([
    [1, complex(1, -1)],
    [complex(1, 1), -1]
])
x = np.array([complex(1, 2), complex(3, 4)])
print(f"U dagger times U:\n{U@U.conj().T}\n")
print(f"Absolute determinant of U:{np.abs(np.linalg.det(U))}\n")
print(f"Norm x : {np.linalg.norm(x, 2)}")
print(f"Norm Ux: {np.linalg.norm(U@x, 2)}\n")
print(f"Eigenvalues: {np.linalg.eigvals(U)}")
```
