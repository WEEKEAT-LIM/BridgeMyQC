# Orthonormal Basis
Properties of orthonormal basis:
- Independent ($Ax=0$, the only solution is $x=0$)
- Orthogonal (dot product with other vector is 0)
- Unit length (magnitude is 1) which means dot product of it self is 1

Example
$$
I = \begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
Columns vectors of $I$ are independent, orthogonal and also have unit length!

```python
import numpy as np
from bridge_my_qc.helpers.linear_algebra import get_column_vectors
from bridge_my_qc.helpers.plot import plot_cartesian_3d
```

```python
I = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
])
column_vec = get_column_vectors(I)


print(f"v1 dot v2: {column_vec[0].dot(column_vec[1])}")
print(f"v1 dot v3: {column_vec[0].dot(column_vec[2])}")
print(f"v3 dot v2: {column_vec[2].dot(column_vec[1])}")
print(
    f"length: v1={np.linalg.norm(column_vec[0])} v2={np.linalg.norm(column_vec[1])} v3={np.linalg.norm(column_vec[2])}"
)
```

## Operation of Orthonormal Bases
Let $Q_{3 \times 3} = \begin{bmatrix} q_1 & q_2 & q_3\end{bmatrix}$, $q_x$ be column vectors.
You can prove the following relationship between $Q$ and $Q^T$ using previous knowledge.

$Q^TQ=I$:
$$
\begin{align}
&\begin{bmatrix} q_1 & q_2 & q_3\end{bmatrix}^T \begin{bmatrix} q_1 & q_2 & q_3\end{bmatrix} \\
&= \begin{bmatrix} q_1 \\ q_2\\ q_3\end{bmatrix} \begin{bmatrix} q_1 & q_2 & q_3\end{bmatrix} \\
&= 
\begin{bmatrix}
q_1 \cdot q_1 & q_1 \cdot q_2 & q_1 \cdot q_3 \\
q_2 \cdot q_1 & q_2 \cdot q_2 & q_2 \cdot q_3 \\
q_3 \cdot q_3 & q_3 \cdot q_2 & q_3 \cdot q_3 \\
\end{bmatrix} \\
&= \begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
\end{align}
$$



$Q^T=Q^{-1}$:
$$
\begin{align}
Q^TQ &= I \\
Q^TQQ^{-1} &= I Q^{-1} \\
Q^TI &= Q^{-1} \\
Q^T &= Q^{-1}
\end{align}
$$

Other Example Q:
$$
\begin{bmatrix}
\frac{1}{\sqrt{2}} & -\frac{1}{\sqrt{2}} & 0\\
\frac{1}{\sqrt{2}} & \frac{1}{\sqrt{2}} & 0 \\
0 & 0 & 1
\end{bmatrix}
$$
> Note that you may not get exact identity matrix in NumPy, but the value is lie within standard `machine precision limits` (The resultant matrix is approximately equal to identity matrix).
> In numerical computation we use $e^{\dots}$ to represent $10^{\dots}$.

```python
%matplotlib ipympl
Q = np.array(
    [[1 / np.sqrt(2), -1 / np.sqrt(2), 0], 
     [1 / np.sqrt(2), 1 / np.sqrt(2), 0],
     [0, 0, 1]]
)
print(f"Transpose:\n{Q.T}\n")
print(f"Q^T Q: \n{Q.T@Q}\n")
print(f"Inverse Q: \n{np.linalg.inv(Q)}")
plot_cartesian_3d(get_column_vectors(Q), "vector")
```
