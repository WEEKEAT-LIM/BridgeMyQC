# Dirac Notation (Bra-Ket notation)
Bra-Ket notation was created by British physicist `Paul Dirac` in 1939 through his paper **A New Notation for Quantum Mechanics**.
It provide a simpler and cleaner way to represent vector. We use it to represent basis of quantum state.

$\vec{v}= \begin{bmatrix}v_1 \\ \dots \\ v_n\end{bmatrix}$, $\vec{v^*}= \begin{bmatrix}v_1^* \\ \dots \\ v_n^*\end{bmatrix}$, $v \in \mathbb{C}$

- $\ket{v}$ (Ket) represent column vector
- $\bra{v}$ (Bra) represent complex conjugate transpose of the column vector, so it is row vector

> Note that $\ket{x}$, $x$ is just a symbol(can be anything) represent the 'state'(vector)

Since we use $\dagger$ (Dagger) to represent complex conjugate transpose, therefore:
- $\bra{v} = \ket{v}^\dagger = \vec{v}^{*T}$
- $\ket{v} = \bra{v}^\dagger = \vec{v}$

The dot product of vector and its conjugate also can be written as $\braket{v|v}$:
$$
\begin{align}
\vec{v} \cdot \vec{v^*} &= \begin{bmatrix}v_1 \\ \dots \\ v_n\end{bmatrix} \cdot \begin{bmatrix}v_1^* \\ \dots \\ v_n^*\end{bmatrix} \\
&=\sum_{i=n}^n v_i \times v_i^* \\
&= \braket{v|v}

\end{align}
$$
We already know $zz^* = a^2 +b^2$, so the length/magnitude of vector is $||\vec{v}|| = \sqrt{\braket{v|v}}$

> The outer product is $\ket{v}\bra{v}$, you will get a matrix instead of scalar

## Orthonormal Bases in terms of Dirac notation
Recap orthonormal bases:
- Basis must orthogonal to each other
- Basis must have unit length (magnitude = 1)

We can rewrite vector in terms of linear combination of orthonormal bases $\vec{v}  = \ket{v}= c_1\ket{b_1} + c_2\ket{b_2} + \dots + c_n\ket{b_n}$

Example $\vec{v} = \begin{bmatrix} -2 \\ 1\end{bmatrix}$:
$$
\begin{align}
\ket{0} = \begin{bmatrix} 1 \\0 \end{bmatrix} \quad \ket{1} = \begin{bmatrix} 0 \\1 \end{bmatrix} \\ \\
\vec{v} = -2\ket{0} + \ket{1}
\end{align}
$$
Check orthonormal:
- Orthogonality : $\braket{0|1} = 0$
- Unit length : $\braket{0|0} = 1$ & $\braket{1|1} = 1$

## Operations
> In the following section $\ket{\psi}$ and $\ket{\phi}$ are NOT orthonormal bases, instead they are 'resultant state'(combination of them)

We can interpret $\hat{A}\ket{\psi}=\ket{\phi}$ as Matrix $\hat{A}$ transform $\ket{\psi}$ to $\ket{\phi}$.

**Associativity**: fully associative
- $A(B\ket{\psi}) = (AB)\ket{\psi}$
- $ \bra{\psi}(A\ket{\phi}) = (\bra{\psi}A)\ket{\phi}$
- $ (\ket{\psi}\bra{\phi})\ket{x} = \ket{\psi}(\braket{\phi|x})$

**Commutativity**: Not commute
- $\hat{A}\ket{\psi} \ne \ket{\psi}\hat{A}$
- $\ket{\psi}\bra{\phi} \ne \ket{\phi}\bra{\psi}$ 
- $\braket{\psi | \phi} = \braket{\phi | \psi}^*$
