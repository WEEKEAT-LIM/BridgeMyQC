# Orthogonal
If any 2 lines are perpendicular to each other there must be a intersection point of these lines. In vector space, any 2 perpendicular($90\degree$) vectors is possible to not touching each other.

![Orthogonal & Perpendicular](../../Assets/images/orthogonal_perpendicular.png)

The only difference between Fig1 and Fig 2 is the magnitude or the length of the vector, but the angle still $90\degree$ even they don't intersect.

Therefore orthogonal is just the language we used in linear algebra, to describe the vectors are $90\degree$ to each other. There is no big difference between orthogonal and perpendicular.

Since $\vec{v_1} \cdot \vec{v2} = |\vec{v_1}| |\vec{v_2}| \cos \theta$, and $\cos 90\degree = 0$,  dot product of `orthogonal` vectors is 0.

```python
import numpy as np
from bridge_my_qc.helpers.plot import plot_cartesian_3d
v1 = np.array([1, 0, 0])
v2 = np.array([0, 1, 0])
print(np.dot(v1, v2))

plot_cartesian_3d([v1, v2], "vector")
```
