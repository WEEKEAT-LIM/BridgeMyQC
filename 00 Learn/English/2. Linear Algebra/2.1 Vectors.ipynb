## Scalar VS Vector
If a quantity only have magnitude such as mass, energy, temperature it is scalar. If a quantity have magnitude and direction such as force then it is vector.
In high school you already know the cartesian coordinates system: (x, y, z). The only different in vector is, it now become arrow point toward (x, y, z) from the origin.

## Length/Magnitude of vector (Euclidean Norm, $\ell_2$)
We use $||\vec{v} ||$ to represent length of the vector or sometimes it called `Euclidean Distance`.

$$
\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ ... \\ v_n \end{bmatrix}
||\vec{v}|| = \sqrt{\sum_{i=1}^n |v_i|^2}
$$

For example:
$$
\vec{v_1} = \begin{bmatrix} 3 \\ 4\end{bmatrix}
||\vec{v_1}|| = \sqrt{3^2 + 4^2} = 5
$$

> If $|| \vec{v_x} || = 1$, then $\vec{v_x}$ is a `Unit Vector`(also denoted by $\hat{v}$).

Unit vector:
$$
\hat{v} = \frac{v}{||\vec{v}||}
$$


### Addition Info: General Formula of Norm ($\ell_n$)
$$
||\vec{v}||_n = \left(\sum_{i=1}^k |v_i|^n\right)^{\frac{1}{n}}
$$

```python
vector = np.array([3, 4])
l2_norm = np.linalg.norm(vector)
print(f"Length of vector: {l2_norm}")
unit_vector = vector / l2_norm
print(f"Unit Vector: {unit_vector}")
l5_norm = np.linalg.norm(vector, ord=5)
print(f"l5 norm: {l5_norm}")
```

## Addition of Vectors

\begin{align}
\vec{v_1} &= 3\begin{bmatrix} 1 \\ 0 \end{bmatrix} + 4\begin{bmatrix} 0 \\ 1 \end{bmatrix} \\
&= \begin{bmatrix} 3 \\ 0 \end{bmatrix} + \begin{bmatrix} 0 \\ 4 \end{bmatrix} \\

## Dot product / Inner Product
$$
\vec{A} \cdot \vec{B} = \sum_{i=1}^n (a_i \times b_i) = ||\vec{A}|| \, ||\vec{B}|| \cos \theta
$$

Also can express as $||\vec{A}||\cos \theta \, ||\vec{B}|| $

![Vector Dot Product](../../Assets/images/Vector_dot.png)

Example:
$$
\begin{align}
\begin{bmatrix} 1 \\ 2 \end{bmatrix} \cdot \begin{bmatrix} 3 \\ 4 \end{bmatrix} &= \sqrt{1^2 + 2^2} \sqrt{3^2 + 4^2} \cos \theta \\
1 \times 3 + 2 \times 4 &= \sqrt{1^2 + 2^2} \sqrt{3^2 + 4^2} \cos \theta  \\
11 &= 5\sqrt{5} \cos \theta
\end{align}

$$

```python
A = np.array([1, 2])
B = np.array([3, 4])
dot_product = np.dot(A, B)
print(f"A dot B : {dot_product}")
```

What would happen when $\theta$ become $0$? Try to create few pairs of vectors perpendicular to each other and check their dot product.

```python
%matplotlib ipympl
vector1 = np.array([1, 0, 0])
vector2 = np.array([0, 1, 0])
dot_product = np.dot(vector1, vector2)
cos_theta = dot_product / (np.linalg.norm(vector1) * np.linalg.norm(vector2))
angle = np.degrees(np.arccos(cos_theta))

plot_cartesian_3d([vector1, vector2], "vector", label=f"Dot Product={dot_product}; angle in degree={angle}")
plt.show()
```
