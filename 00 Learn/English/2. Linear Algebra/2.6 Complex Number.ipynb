## Background
Recap solving quadratic equation:
$$
x = \frac{-b\pm \sqrt{b^2-4ac}}{2a}
$$
Where $\Delta = b^2-4ac$:
- When $\Delta > 0$: The equation has 2 different real roots
- When $\Delta = 0$: The equation has 1 real root
- When $\Delta < 0$: The equation don't have real root

Therefore, $i = \sqrt{-1}$ was recognized as truly useful at 19th century to solve $\Delta < 0$, but why it takes so long time to recognized the value of complex number? Check out the [history](https://youtu.be/aCEZ853zJWk).

## What is complex number $\mathbb{C}$?
$z = a + bi$, $z \in \mathbb{C}$
- a is real part of the complex number
- b is imaginary part of complex number

Since $i$ can't be found in any real axis, we need to introduced a new axis to plot the complex number (as vector) on the plane (`Complex Plane`).
We use 2 axis to plot a complex number, horizontal represent real part, vertical represent imaginary part.

## Complex Number operations
- **Addition and Subtraction**
  $$
  \begin{align}
    z_1 + z_2 &= (a_1 + b_1 i) + (a_2 + b_2i) \\
    &= (a_1 + a_2) + (b_1 + b_2)i \\ \\
    z_1 - z_2 &= (a_1 + b_1 i) - (a_2 + b_2i) \\
    &= (a_1 - a_2) + (b_1 - b_2)i
  \end{align}
  $$
- **Multiplication**(similar with algebra multiplication)
  $$
  \begin{align}
  z_1 \times z_2 &= (a_1 + b_1 i) \times (a_2 + b_2i) \\
  &= a_1a_2 + (a_1b_2 + a_2b_1)i + (b_1b_2)ii \\
  &= a_1a_2 + (a_1b_2 + a_2b_1)i + (b_1b_2)(-1) \\
  &= (a_1a_2 -b_1b_2)  + (a_1b_2 + a_2b_1)i
  \end{align}
  $$

Example:

$z_1 = 1 + 2i$ ; $z_2 = 3 + 4i$

$$
\begin{align}
  z_1 + z_2 &= (1 + 2 i) + (3 + 4i) \\
  &= (1 + 3) + (2 + 4)i \\
  &= 4 + 6i \\ \\
  z_1 - z_2 &= (1 + 2i) - (3 + 4i) \\
  &= (1 - 3) + (2 - 4)i \\ 
  &= -2 - 2i \\ \\
  z_1 \times z_2 &= (1 + 2i) \times (3 + 4i) \\
  &= (1)(3) + [(1)(4)+(2)(3)]i + (2)(4)i^2 \\
  &= (3-8) + (4+6)i \\
  &= -5 + 10i
\end{align}
$$


> Magnitude from vector view:
> $|z_1z_2 | = |z_1| \, |z_2|$



````python
# Python use j to represent i, because programmer like to use i to represent index XD
z1 = 1 + 2j
z2 = 3 + 4j
mult = z1 * z2

print(f"Addition: {z1 + z2}")
print(f"Subtraction: {z1-z2}")
print(f"Multiplication: {mult}")

# Magnitude comparison
abs(mult) == abs(z1)*abs(z2)
````

- **Conjugation**
  
  If $z = a + bi$, its conjugate ($\bar{z}$ or $z^*$) is $a -bi$, $|z| = |z^*|$

  $$
  \begin{align}
  z^* \times z &= (a - bi) \times (a + bi) \\
  &= a^2 - (bi)^2 \\
  &= a^2 + b^2
  \end{align}
  $$
  The multiplication with conjugate will let imaginary part disappear

  Wait a minute! If $|a|=|b|$, then we plot $z$ and $z^*$ in the complex plane, what is the angle between them? ($90\degree$)

  Let check it using vector's dot product, we define $z= 5 + 5i$
  $$
  \begin{bmatrix} 5 \\5 \end{bmatrix} \cdot \begin{bmatrix} 5 \\ -5 \end{bmatrix} = 5^2 - 5^2 = 0
  $$

````python
z = complex(2, -9)
z_conj = z.conjugate()
print(z_conj)

# Note that you also can get specific part from complex complex number
z_conj.real, z_conj.imag
````

## Polar Coordinates
In Cartesian Coordinate System we mark a point/vector as (x, y). In Polar Coordinate system be mark a point as $(r, \theta)$, r is radius(distance between point and origin), $\theta$ is angle between the r and initial line(x-axis or $\theta=0$).

- We express polar equation in terms of $r$ and $\theta$
- We express cartesian equation in terms of x and y
- We can convert them using trigonometry function and pythagoras ($r^2 = x^2 + y^2$)
- No matter you plot line or curve of equation on (x,y) graph or $(r, \theta)$ , they are always same(exactly fit to each other)

Example:
$$
\begin{align}
r &= 2\cos \theta \\
r &= 2 \frac{x}{r} \\
r^2 &= 2x \\
x^2 +y^2 &= 2x
\end{align}
$$

## Euler formula
> Euler's number(not Euler's constant, they are different) $e \approx 2.71828 \dots$
$$
e^{i\theta}= \cos \theta + i \sin \theta 
$$

Let prove this equation:
1. According to Figure 1: $z = |z| \cos \theta + i |z|\sin \theta$
2. When it is unit circle in complex plane, $|z| = 1 = \sqrt{\cos^2 \theta + \sin^2 \theta}$
3. Now $z = \cos \theta + i \sin \theta $
4. Expand right hand side using Taylor Series:
   $$
   \begin{align}
    &\cos \theta + i \sin \theta \\
    &= (1-\frac{\theta^2}{2!}+\frac{\theta^4}{4!}-\frac{\theta^6}{6!}+ \dots) + i(\theta -\frac{\theta^3}{3!}+\frac{\theta^5}{5!}-\frac{\theta^7}{7!} + \dots) \\
    \\
    & i^1=i \quad i^2 = -1 \quad i^3= -i \quad i^4 = 1 \quad i^5 = i \quad \dots \\ \\
    &= (1+\frac{(i\theta)^2}{2!}+\frac{(i\theta)^4}{4!}+\frac{(i\theta)^6}{6!}+ \dots) + (i\theta +\frac{(i\theta)^3}{3!} +\frac{(i\theta)^5}{5!} +\frac{(i\theta)^7}{7!}) \\
    &= 1 + i\theta +\frac{(i\theta)^2}{2!} +\frac{(i\theta)^3}{3!}+\frac{(i\theta)^4}{4!}+ \dots \\
    &= e^{i\theta}
   \end{align}
   $$

> When $\theta = \pi$, $e^{i\pi}= \cos \pi + i \sin \pi = -1 + 0i =-1 $. Hence, $e^{i\pi}+1 =0$.

````python
import numpy as np
import matplotlib.pyplot as plt
from bridge_my_qc.helpers.plot import plot_complex_vectors
from bridge_my_qc.helpers.linear_algebra import get_column_vectors
````

````python
%matplotlib ipympl
thetas = np.linspace(0, 4*np.pi, 300)
imaginary = np.sin(thetas)
real = np.cos(thetas)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection="3d")


ax.plot(
    thetas,
    real,
    imaginary,
    label=r"Euler's Formula: $e^{it} = \cos(t) + i\sin(t)$",
    color="b",
    linewidth=2,
)

ax.set_xlabel(r"Angle ($\theta$)")
ax.set_ylabel(r"Real ($\cos \theta$)")
ax.set_zlabel(r"Imaginary ($\sin \theta$)")
ax.set_title("3D Plot of Euler's Formula")

ax.legend()
plt.show()
````

### Rotation in complex plane
When we try to rotate $z_1 = 1 + 2i$ by  $90\degree$ or $\frac{\pi}{2}$:
$$
\begin{align}
e^{i\frac{\pi}{2}}(1+ 2i) &= (\cos \frac{\pi}{2} + i \sin \frac{\pi}{2})(1 +2i) \\
&= (0+i)(1+2i) \\
&= -2 + i
\end{align}
$$

## Matrix version Complex Number 


In matrix section we do mention that matrix is also an operator, so can we create complex version rotation($e^{i\theta}$) matrix? (YES!)
$$
e^{i\theta} = \cos \theta \, I + \sin \theta \, I^\prime
$$
Since identity matrix is equivalent to 1, We can just use identity to take combination of real part, what about imaginary part? $i^2=-1$, we need a real matrix behave like imaginary part $(I^\prime)^2 = -I$
$$
\begin{align}
I^\prime &= \begin{bmatrix} 0 & -1 \\ 1 & 0\end{bmatrix} \\
(I^\prime)^2 &= \begin{bmatrix} -1 & 0 \\ 0 & -1\end{bmatrix}
\end{align}
$$

$$
\begin{align}
e^{i\theta} &= \cos \theta \, I + \sin \theta \, I^\prime \\
&= \begin{bmatrix} \cos \theta & 0 \\ 0 & \cos \theta\end{bmatrix} + \begin{bmatrix}  0 & -\sin \theta \\ \sin \theta & 0\end{bmatrix} \\
&= \begin{bmatrix} \cos \theta & -\sin \theta \\ \sin \theta & \cos \theta\end{bmatrix}
\end{align}
$$
This is exactly anticlockwise rotation matrix!

````python
I = np.array([
    [0, -1],
    [1, 0]
])
I@I
````

How to represent complex number in matrix form in order to apply rotation in complex plane?
Since $z = a + bi$, 1 equivalent to $I$ , $i$ equivalent to $I^\prime$:
$$
\begin{align}
z &= a + bi \\
&= aI +bI^\prime \\
&= \begin{bmatrix} a & -b \\ b & a\end{bmatrix}
\end{align}
$$
Hence:
$$
\begin{align}
z_r = e^{i\theta}z &= \begin{bmatrix} \cos \theta & -\sin \theta \\ \sin \theta & \cos \theta\end{bmatrix} \begin{bmatrix} a & -b \\ b & a\end{bmatrix} \\
&= 
\begin{bmatrix} a \cos \theta -b \sin \theta & -b \cos \theta -a \sin \theta \\
a \sin \theta + b \cos \theta & -b \sin \theta + a \cos \theta
\end{bmatrix} \\
&= \begin{bmatrix} a_r & -b_r\\
b_r & a_r
\end{bmatrix}
\end{align}
$$
Rotated $z = (a \cos \theta -b \sin \theta) + (a \sin \theta + b \cos \theta)i$

````python
z = 1 + 2j

Z = np.array([
    [z.real, -z.imag],
    [z.imag, z.real]
])

def anticlockwise_rotation_matrix(degrees):
    theta = np.radians(degrees)
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c, -s], [s, c]])

acR = anticlockwise_rotation_matrix(90)
cR = anticlockwise_rotation_matrix(-90)

plot_complex_vectors(
    [
        z,
        complex(*get_column_vectors(acR @ Z)[0]),
        complex(*get_column_vectors(cR @ Z)[0]),
    ]
)
````
