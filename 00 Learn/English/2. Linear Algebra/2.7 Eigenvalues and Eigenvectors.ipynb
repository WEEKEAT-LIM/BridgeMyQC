# Eigenvalues and Eigenvectors
$$
A\vec{x}= \lambda \vec{x}
$$
$A$ transform $\vec{x}$ to vector having same direction with different scale. $x$ is Eigenvector, $\lambda$ is Eigenvalue (the scale, can be negative).
![Eigenvector and Eigenvalue](../../Assets/images/eigen1.png)

- Matrix $A$ can have 2 or more Eigenvectors and corresponding Eigenvalues
- $\sum \lambda = tr(A)$ and $\Pi \, \lambda = \det A$
- Zero vector is not an Eigenvector, $x \ne 0$
- When $A$ has different Eigenvalues, the corresponding Eigenvectors are linearly independent and become bases to span a vector space

## Find out Eigenvalues and Eigenvectors
> Note that when $AB = X$, $\det A \times \det B = \det X$
$$
\begin{align}
A\vec{x} &= \lambda \vec{x} \\
A\vec{x} - \lambda \vec{x} &= 0 I\\
(A - \lambda I) \vec{x}&= 0 I\\
\det{(A - \lambda I)} &= 0 \\
\end{align}
$$

Example $A = \begin{bmatrix} 6 &2 \\ 4 & 8\end{bmatrix}$:

Step 1: Find Eigenvalues
$$
\begin{align}
\det(\begin{bmatrix} 6 &2 \\ 4 & 8\end{bmatrix} - \lambda \begin{bmatrix} 1 &0 \\ 0 & 1\end{bmatrix}) &= 0\\
\det(\begin{bmatrix} 6-\lambda & 2 \\ 4 & 8-\lambda\end{bmatrix}) &= 0 \\
(6-\lambda)(8-\lambda) - (4)(2) &= 0 \\
\lambda^2 -14\lambda + 48 -8 &= 0 \\
\lambda^2 -14\lambda + 40 &= 0 \\
(\lambda -4)(\lambda - 10) &= 0 \\
\lambda \in \set{4, 10}
\end{align}
$$
Step 2: Find Eigenvectors
$$
\begin{align}
(A - \lambda I) \vec{x}&= 0 \\
\begin{bmatrix} 6-4 & 2 \\ 4 & 8-4\end{bmatrix}\vec{x} &= 0 \\
\begin{bmatrix} 2 & 2 \\ 4 & 4\end{bmatrix}\vec{x} &= 0 \\
\vec{x} &= c\begin{bmatrix} 1 \\ -1 \end{bmatrix}, c\in \mathbb{R}
\end{align}
$$

$$
\begin{align}
(A - \lambda I) \vec{x}&= 0 \\
\begin{bmatrix} 6-10 & 2 \\ 4 & 8-10\end{bmatrix}\vec{x} &= 0 \\
\begin{bmatrix} -4 & 2 \\ 4 & -2\end{bmatrix}\vec{x} &= 0 \\
\vec{x} &= c\begin{bmatrix} 1 \\ 2 \end{bmatrix}, c\in \mathbb{R}
\end{align}
$$

Therefore, A has Eigenvalue of 4, 10 with corresponding Eigenvectors $\begin{bmatrix} 1 \\ -1 \end{bmatrix}$, $\begin{bmatrix} 1 \\ 2 \end{bmatrix}$. Usually we will also normalize the Eigenvectors (get the unit vectors, $\vec{u_x}=\frac{\vec{x}}{||\vec{x}||}$).

> Note that NumPy output normalized eigenvectors

```python
import numpy as np
import matplotlib.pyplot as plt
from bridge_my_qc.helpers.plot import plot_cartesian_2d
from bridge_my_qc.helpers.linear_algebra import get_column_vectors
```

```python
A = np.array([
    [6, 2],
    [4, 8]
])
X = np.array([[1, 1], [-1, 2]])
lx = A@X
vectors = []
vectors.extend(get_column_vectors(X))
vectors.extend(get_column_vectors(lx))
plot_cartesian_2d(vectors, 'vector')
plt.show()
```

```python
eigen_value, normalized_eigen_vectors = np.linalg.eig(A)
print(
    f"Eigenvalues:{eigen_value}\nEigenvectors(normalized):\n{normalized_eigen_vectors}"
)
vectors = []
vectors.extend(get_column_vectors(normalized_eigen_vectors))
vectors.extend(get_column_vectors(A@normalized_eigen_vectors))
plot_cartesian_2d(vectors, "vector")
plt.show()
# You may see vectors in opposite direction to previous graph, but they are in the same line(still considered as eigenvectors)
```

Example2: $A=\begin{bmatrix}2&0 \\-1&2 \end{bmatrix}$ (when has same eigenvalues)

Step 1: Find eigenvalue
$$
\begin{align}
\det(\begin{bmatrix}2-\lambda & 0 \\ -1 & 2-\lambda \end{bmatrix}) = 0 \\
(2-\lambda)(2-\lambda) = 0 \\
\lambda =2
\end{align}
$$
Step 2: Find eigenvector
$$
\begin{align}
\begin{bmatrix}2-2 & 0 \\ -1 & 2-2 \end{bmatrix} \vec{x} &= 0 \\
\begin{bmatrix}0 & 0 \\ -1 & 0 \end{bmatrix} \vec{x} &= 0 \\
\vec{x} &= \begin{bmatrix} 0 \\ c\end{bmatrix}\
\end{align}
$$
In this case $c$ is free variable, we can set it to 1(convenient for further calculation).

```python
A = np.array([
    [2, 0],
    [-1, 2]
])
np.linalg.eig(A)
```

Example 3: $A=\begin{bmatrix}0&-1 \\1&0 \end{bmatrix}$ (rotation matrix of $90 \degree$ has complex eigenvalues)

$\det A = 1$ and $tr(A)=0$

$$
\begin{align}
\det(\begin{bmatrix}0-\lambda & -1 \\ 1 & 0-\lambda\end{bmatrix}) &= 0 \\
(-\lambda)^2 +1&=0 \\
\lambda^2 +1 &= 0 \\
(\lambda + i)(\lambda - i) &= 0 \\
\lambda &\in \set{-i, i}
\end{align}
$$

$$
\begin{align}
\begin{bmatrix}i & -1 \\ 1 & i\end{bmatrix}\vec{x_1} &= 0 &\quad \begin{bmatrix}-i & -1 \\ 1 & -i\end{bmatrix}\vec{x_2} &=0\\
\vec{x_1} &= \begin{bmatrix}1\\i\end{bmatrix} &\quad \vec{x_2}&=\begin{bmatrix}i\\1\end{bmatrix}
\end{align}
$$

```python
A = np.array([[0, -1], [1, 0]])
np.linalg.eig(A)
```

## Diagonalizing Matrix
When $A_{n \times n}$ has $n$ Eigenvalues and $n$ Eigenvectors:
$$
\begin{align}
X = \begin{bmatrix} x_1 & \dots & x_n\end{bmatrix} &\quad \Lambda = diag(\lambda_1, \lambda_2, \dots, \lambda_n) \\
X^{-1}AX &=\Lambda  \\
A &= X \Lambda X^{-1}
\end{align}
$$
This will make our life easier, you don't need find $A^{k}$ by doing k times matrix multiplication, instead:
$$
A = X \Lambda X^{-1} \\
A^{k} = (X \Lambda X^{-1})^k = (X \Lambda X^{-1})(X \Lambda X^{-1}) \dots (X \Lambda X^{-1}) \\
A^k = X \Lambda^k X^{-1}
$$
Since $\Lambda$ is diagonal matrix:
$$
\begin{align}
\Lambda_{3 \times 3} &= \begin{bmatrix} \lambda_1 & 0 & 0 \\ 0 & \lambda_2 & 0 \\ 0&0&\lambda_3\end{bmatrix} \\
\Lambda^k &= \begin{bmatrix} \lambda_1^k & 0 & 0 \\ 0 & \lambda_2^k & 0 \\ 0&0&\lambda_3^k\end{bmatrix}
\end{align}
$$

> Note that column vectors of $\Lambda$ (Eigenstates) are linearly independent and orthogonal to each others 

```python
A = np.array([
    [8, 3],
    [2, 7]
])

lambdas, X = np.linalg.eig(A)
Lambda = np.diag(lambdas)
X_inv = np.linalg.inv(X)


A2 = A@A@A@A@A
A3 = X@(Lambda**5)@X_inv

print(X@Lambda@X_inv)
print(A2)
print(A3)
```
