Given any $\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ v_3\end{bmatrix}$, we can plot it in a 3D space. How many $\vec{v}$ can we plot in 3D space?(infinite) Basis is fundamental building block of the space, like LEGO! If I give you finite different lego blocks, you can use them to build a lot of different things. So, the lego blocks determine what are you able to build. Or you can think basis as your custom axes for your custom space.Which mean:
- number of basis is `dimension` of the space
- basis must be `independent`

## Independent
If we change the matrix from previous example to $A = \begin{bmatrix}1&0&-5 \\ 0&1&3 \\ 0&0&0\end{bmatrix}$, it now jus span a **plane** even we have 3 different column vectors and each vector have 3 component. If we plot all combinations of the vectors of A, you will see they all lie in a plane, because none of these vectors have z-component, so you are not able to build vector have z-component. All z-component of vectors in the space is 0, which also mean you can plot all of them without including z-axis in your graph. This is because ONLY 2 of 3 column vectors in A are `independent`, means A only have 2 `basis`.

>**What is `independent` meant?**
>
>When $\vec{v_1}$, $\vec{v_2}$, and $\vec{v_3}$ are independent, mean you are not able to get any 1 vector by taking combination of other vectors.
>
> They not necessary perpendicular to each other, but they definitely in different direction.

### How to check whether vectors are independent?
Lets review matrix A from column perspective:
$$
\vec{v_1} = \begin{bmatrix}1 \\ 0 \\ 0\end{bmatrix} \,
\vec{v_2} = \begin{bmatrix}0 \\ 1 \\ 0\end{bmatrix} \,
\vec{v_3} = \begin{bmatrix}-5 \\ 3 \\ 0\end{bmatrix}
$$

$$
\begin{align}
\vec{v_3} = -5\vec{1} +3 \vec{v_2}
\end{align}
$$
So
$$
\begin{align}
Ax &= 0 \\
\begin{bmatrix}1&0&-5 \\ 0&1&3 \\ 0&0&0\end{bmatrix}\begin{bmatrix}-5 \\3 \\ -1\end{bmatrix} &=0
\end{align}
$$

If $x_o =$ any combination of possible $x$, you will find out $Ax_{0} = 0$ 

So we can say ALL the column vectors of A are independent when $Ax=0$ ONLY has 1 solution which is $x=0$

```python
# We can also check the determinant of A is also zero
A = np.array([
    [1, 0, -5],
    [0, 1, 3],
    [0, 0, 0]
])
np.linalg.det(A)
```

We also can conclude that if vectors of matrix are not independent to each other, mean the matrix must be `singular`.

```python
x = np.array([
    [-5],
    [3],
    [-1], 
])
A@x
```
