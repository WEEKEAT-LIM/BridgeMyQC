# Spaces
In previous sections, all matrix operation are operate within same global space(universal dimension). The following section is about different ways to constructing/combining spaces.


| Approaches | Notation | Dimension| Purpose |
| -- | -- | --| --|
| Direct Sum | $V \bigoplus W$ | dim(V) + dim(W) | Combine independent spaces without cross-interaction|
| Dual Space | $V^*$ | dim(V) | asdas |
|Tensor Product | $V \bigotimes W$ | dim(V) $\times$ dim(W) | space multiplication

## Direct Sum
$V$(space) has basis(spanned by): $b_v = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix}$, $v \in V$

$W$(space) has basis(spanned by): $b_w = \begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix}$, $w \in W$

$$
\begin{align}
z &= 1\begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix} + 3\begin{bmatrix} 0 \\ 1 \\ 1 \end{bmatrix} \\
&= \begin{bmatrix} 1 \\ 3 \\ 3 \end{bmatrix}
\end{align}
$$

>QUESTION: 
>It is clear that z is linear combination of v and w, z related with V and W. What space z belong to, and how to represent the space in terms of V and W?
>- $z \notin V \cup W$, because it neither factor of $b_v$ nor $b_w$
>- $V \cap W$ = $\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$ ONLY

Therefore we use `Direct Sum` to represent this relation ship:
$$
\begin{align}
z &\in (V \bigoplus W) \\ \\
dim(z) &= dim(v) + dim(w) \\
&= 1+2
\end{align}
$$

Note that if $V \cap W$ have other solution than $\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$, W and V not able to have direct sum.

## Tensor Product (Kronecker Product)
The idea is multiply the space by 'broadcast' the space to other. 'Similar' to scalar matrix multiplication, just change scalar to matrix.

Example:
$$
\begin{align}
V = \begin{bmatrix}1&1\\1&1\end{bmatrix} &\quad W = \begin{bmatrix}2&4\\6&8\end{bmatrix} \\ \\
W \bigotimes V &= \begin{bmatrix}2\begin{bmatrix}1&1\\1&1\end{bmatrix} &4\begin{bmatrix}1&1\\1&1\end{bmatrix} \\6\begin{bmatrix}1&1\\1&1\end{bmatrix} &8\begin{bmatrix}1&1\\1&1\end{bmatrix} \end{bmatrix} \\
&= \begin{bmatrix} 2 &2&4&4 \\2 &2&4&4 \\6 &6&8&8 \\6 &6&8&8 \end{bmatrix}
\end{align}
$$

```python
import numpy as np
```

```python
v = np.array([
    [1, 1],
    [1, 1]
])
w = np.array([
    [2, 4],
    [6, 8]
])

print(f"W tensor V:\n{np.kron(w, v)}\n")
print(f"V tensor W:\n{np.kron(v, w)}")
```
